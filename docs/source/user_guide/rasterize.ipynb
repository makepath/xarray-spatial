## Rasterize

`xrspatial.rasterize` converts vector geometries (polygons, lines, points) into a 2D `xr.DataArray`. No GDAL dependency required.

This guide covers:
- [Basic rasterization](#Basic-rasterization) -- polygons, lines, and points
- [Merge modes](#Merge-modes) -- controlling how overlapping geometries combine
- [Custom merge functions](#Custom-merge-functions) -- user-defined numba-jitted merge logic
- [Dask parallel rasterization](#Dask-parallel-rasterization) -- tile-based output chunking
